# Number of cores per context domain for June-September (2004-2024)

```python
import numpy as np                                                                      
import matplotlib.pyplot as plt                                                         
```

```python
nb_cores_lt_long = np.load("../../output/x0-analysis/nbx0-dakar-long.npy")
nb_cores_lt_short = np.load("../../output/x0-analysis/nbx0-dakar-short.npy")
```

# Short lead time

```python
plt.hist(nb_cores_lt_short, bins=10, color="red", alpha=0.7)
plt.xlabel(f"Number of cores")
plt.ylabel("Frequency")
plt.title(f"Short lead time")
plt.show()
```

```python
np.median(nb_cores_lt_short)
```

# Long lead time

```python
plt.hist(nb_cores_lt_long, bins=20, color="green", alpha=0.7)
plt.xlabel(f"Number of cores")
plt.ylabel("Frequency")
plt.title(f"Long lead time")
plt.show()
```

```python
np.mean(nb_cores_lt_long)
```
